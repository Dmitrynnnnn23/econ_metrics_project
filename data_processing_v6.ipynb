```python
def make_dataset(data_dirty):
    rows = []

    for i in range(data_dirty.shape[0]):
        def get_json(column_name, index):
            val = data_dirty[column_name][index]
            if isinstance(val, str) and val.strip():
                try:
                    return json.loads(val)
                except:
                    return None
            return None

        try:
            ## скидка
            discount_data = json.loads(data_dirty['pinfs'][i])['skidka']
            discount = None if not discount_data else discount_data['values'][0]['key']

            ## количество альбомов
            cur_album = json.loads(data_dirty['albums'][i])
            album_count = (cur_album['totalCount'])

            ## количество фоток
            cur_files = json.loads(data_dirty['files'][i])
            photos_count = (cur_files['totalCount'])

            # Проверка паспорта
            badges = get_json('badges', i)
            docs_verified = (badges[0]["type"] == 'verified') if (badges and len(badges) > 0) else False

            # Пол
            gender = data_dirty['gender'][i] if pd.notna(data_dirty['gender'][i]) else None

            # Гарантия
            guarantee_data = get_json('guaranteeBlock', i)
            master_garantee = None
            if guarantee_data and 'content' in guarantee_data and len(guarantee_data['content']) > 1:
                master_garantee = guarantee_data['content'][1].get('template')
            master_garantee = 'Нет гарантии' if not master_garantee else master_garantee
            if master_garantee.split()[0] == 'Советуем':
                master_garantee = 'По договоренности'
            if master_garantee == '«Сделаю скидку»':
                master_garantee = 'По договоренности'

            # Рейтинг и отзывы
            mean_rating = data_dirty['newRank'][i] if pd.notna(data_dirty['newRank'][i]) else 0
            reviews_count = data_dirty['reviewsCount'][i] if pd.notna(data_dirty['reviewsCount'][i]) else 0

            # json с образование и опыт - сложно парсить плюс нет фиксированных категорий - просто закинем в LLM
            experience_education =  data_dirty['assembledInfoListing'][i]

            # Отметка "очень хвалят"
            trust_badges = get_json('trustBadges', i)
            mark_hvalyat = (trust_badges[0]['type'] == 'great') if (trust_badges and len(trust_badges) > 0) else False

            # Проверка квалификации
            cert_data = get_json('certifiedServices', i)
            work_cert = (cert_data.get('title') == 'Квалификация подтверждена') if cert_data else False

            ## id мастера
            master_id = data_dirty.iloc[i, 1]

            # Цена
            ## делаем нормировку на минимальную цену другой похожей категории цены
            price_list = get_json('priceListPreview', i)
            socket_price = None
            if price_list and 'prices' in price_list:
                for item in price_list['prices']:
                    if item.get('price', {}).get('name').lower() == 'установка розеток и выключателей':
                        socket_price = (item['price'].get('from'))
                        norm_coef = 1
                        if socket_price:
                            socket_price = float(socket_price) / norm_coef
                            break

                    if item.get('price', {}).get('name').lower() == 'установка розеток':
                        norm_coef = 1.5
                        socket_price = (item['price'].get('from'))
                        if socket_price:
                            socket_price = float(socket_price) / norm_coef
                            print('установка розеток')
                            break

                    if item.get('price', {}).get('name').lower() == 'установка розетки':
                        norm_coef = 1
                        socket_price = (item['price'].get('from'))
                        if socket_price:
                            socket_price = float(socket_price) / norm_coef
                            print('установка розетки')
                            break

                    if item.get('price', {}).get('name').lower() == 'ремонт розеток':
                        norm_coef = 1
                        socket_price = (item['price'].get('from'))
                        if socket_price:
                            socket_price = float(socket_price) / norm_coef
                            print('ремонт розеток')
                            break

                    if item.get('price', {}).get('name').lower() == 'замена розетки':
                        norm_coef = 1.55
                        socket_price = (item['price'].get('from'))
                        if socket_price:
                            socket_price = float(socket_price) / norm_coef
                            print('замена розетки')
                            break

                    if item.get('price', {}).get('name').lower() == 'перенос розетки':
                        norm_coef = 1
                        socket_price = (item['price'].get('from'))
                        if socket_price:
                            socket_price = float(socket_price) / norm_coef
                            print('перенос розетки')
                            break

                    if item.get('price', {}).get('name').lower() == 'установка наружных розеток':
                        norm_coef = 0.9
                        socket_price = (item['price'].get('from'))
                        if socket_price:
                            socket_price = float(socket_price) / norm_coef
                            print('установка наружных розеток')
                            break

                    if item.get('price', {}).get('name').lower() == 'установка внутренних розеток':
                        norm_coef = 0.9
                        socket_price = (item['price'].get('from'))
                        if socket_price:
                            socket_price = float(socket_price) / norm_coef
                            print('установка внутренних розеток')
                            break

                    if item.get('price', {}).get('name').lower() == 'установка розетки с заземлением':
                        norm_coef = 0.9
                        socket_price = (item['price'].get('from'))
                        if socket_price:
                            socket_price = float(socket_price) / norm_coef
                            print('установка розетки с заземлением')
                            break

                    if item.get('price', {}).get('name').lower() == 'установка умной розетки':
                        norm_coef = 1
                        socket_price = (item['price'].get('from'))
                        if socket_price:
                            socket_price = float(socket_price) / norm_coef
                            print('установка умной розетки')
                            break

                    if item.get('price', {}).get('name').lower() == 'установка уличной розетки':
                        norm_coef = 1
                        socket_price = (item['price'].get('from'))
                        if socket_price:
                            socket_price = float(socket_price) / norm_coef
                            print('установка уличной розетки')
                            break

                    if item.get('price', {}).get('name').lower() == 'установка розеток в столешницу':
                        norm_coef = 1
                        socket_price = (item['price'].get('from'))
                        if socket_price:
                            socket_price = float(socket_price) / norm_coef
                            print('установка розеток в столешницу')
                            break

                    if item.get('price', {}).get('name').lower() == 'установка розеток на кухне':
                        norm_coef = 1.5
                        socket_price = (item['price'].get('from'))
                        if socket_price:
                            socket_price = float(socket_price) / norm_coef
                            print('установка розеток на кухне')
                            break

                    if item.get('price', {}).get('name').lower() == 'установка блока розеток':
                        norm_coef = 1.5
                        socket_price = (item['price'].get('from'))
                        if socket_price:
                            socket_price = float(socket_price) / norm_coef
                            print('установка блока розеток')
                            break

                    if item.get('price', {}).get('name').lower() == 'электрика':
                        norm_coef = 1
                        socket_price = (item['price'].get('from'))
                        if socket_price:
                            socket_price = float(socket_price) / norm_coef
                            print('электрика')
                            break

                    if item.get('price', {}).get('name').lower() == 'aварийный выезд электрика':
                        norm_coef = 3
                        socket_price = (item['price'].get('from'))
                        if socket_price:
                            socket_price = float(socket_price) / norm_coef
                            print('aварийный выезд электрика')
                            break
            ## подсказки от профи ру
            cur_hints = json.loads(data_dirty['suitabilityHints'][i])
            suit_hint = None if not cur_hints else cur_hints[0]['text']

            ## есть ли договор и страховка от профи ру
            cur_badges = json.loads(data_dirty['badges'][i])
            insurance = None
            contract = None

            for badge in cur_badges:
                if badge['type'] == 'insurance':
                    insurance = badge['text']
                if badge['type'] == 'contract':
                    contract = badge['text']

            ## хакнули скоры от профи ру
            cur_scores = json.loads(data_dirty['scores'][i])
            cur_score = cur_scores['score']

            ## когда мастер зарегистровался на профи ру
            cur_listing = json.loads(data_dirty['assembledInfoListing'][i])
            profi_ru_start = None
            for data in cur_listing:
                if data['type'] == 'INFO':
                    cur_fact = data['oioFact']
                    if "На сервисе с"in cur_fact['fact']['title']:
                        profi_ru_start = (cur_fact['fact']['title'])

            # Сборка данных строки
            row = {
                "master_id": master_id,
                "docs_verified": docs_verified,
                "gender": gender,
                "guarantee": master_garantee,
                "rating": mean_rating,
                "reviews_count": reviews_count,
                "experience_education": experience_education,
                "is_highly_praised": mark_hvalyat,
                "is_certified": work_cert,
                "price_socket": socket_price,
                "suitabilityHints": suit_hint,
                "discount": discount,
                "insurance": insurance,
                "contract": contract,
                "cur_score": cur_score,
                "profi_ru_start": profi_ru_start,
                "album_count": album_count,
                "photos_count": photos_count
            }
            rows.append(row)

        except Exception as e:
            print(f"Критическая ошибка на строке {i}: {e}")
            rows.append({})
    return pd.DataFrame(rows)
```

```python
def label_data(text: str, giga: GigaChat):
    """Классификация текста через GigaChat."""

    # Формируем четкую инструкцию для модели
    prompt = ( """Ты опытный менеджер по подбору мастеров по установке розеток в городе Казань, отлично умеешь оценивать уровень образования и опыт работы мастеров.

#Задача
Оценить уровнь образования мастера и его опыт работы по описанию.

Уровень образования (в формате True/False - имеется ли у мастера образование этого уровня):
1. Профессиональное среднее образование: если мастер закончил техникум, коледж, ПТУ и т.д. включая аббревиатуры конкретных мест обучения, относящихся к профессиональному среднему образованию
2. Высшее образование: если мастер закончил университет, институт и т.д., включая аббревиатуры конкретных мест обучения, относящихся к высшему образованию


Опыт работы (в формате число или None):
Необходимо выделить сумму лет работы мастера. Учитывай, что сейчас 2026 год. То есть "опыт работы с 2021 года" означает 5 лет опыта работы. Верни только число лет работы.
Если опыт работы не указан, верни None

# Формат вывода

Выводи результат строго в формате JSON следующего вида

{'Профессиональное среднее образование': True / False,
 'Высшее образование': True / False,
 'Опыт работы': суммарное число лет работы (только число)}

# Пример:
Описание мастера:
'Казанский энергетический техникум, 2011–2014 гг. ИНК-Сервис занимался монтажом и обслуживанием электрооборудования, ремонтировал сантехнику , мебель , чистил снег, 2018–2025 гг.'

Правильный вывод:

{'Профессиональное среднее образование': True,
 'Высшее образование': False,
 'Опыт работы': 8}

Верни корректный JSON для описания мастера: """
f'{text}')
    try:
        response = giga.chat(prompt)
        content = response.choices[0].message.content

        # Находим JSON в ответе (на случай, если модель добавит лишний текст)
        start = content.find('{')
        end = content.rfind('}') + 1
        return json.loads(content[start:end])
    except Exception as e:
        return {"category": "Ошибка", "details": str(e)}
```

```python
print("🚀 Начинаю разметку через GigaChat...")
dataset = list(all_data_final['experience_education'])
results = []

# verify_ssl_certs=False критичен, если у вас нет сертификатов Минцифры
with GigaChat(credentials=AUTH_KEY, verify_ssl_certs=False) as giga:
    for i, text in enumerate(dataset, 1):
        print(f'Обработка документа {i}...')
        label = label_data(text, giga)
        label['master_id'] = all_data_final.iloc[i - 1, 0]
        results.append(label)
```
