```python
def profiru_start_transformer(value):
    months = {'января': 1, 'февраля': 2, 'марта': 3, 'апреля': 4, 'мая': 5, 'июня': 6,
        'июля': 7, 'августа': 8, 'сентября': 9, 'октября': 10, 'ноября': 11, 'декабря': 12}

    if pd.isna(value) or not isinstance(value, str):
        return None

    match = re.search(r"(?:На сервисе с|На Профи\.ру с)\s+([а-яё]+)\s+(\d{4})", value, re.IGNORECASE)
    if match:
        month_str = match.group(1).lower()
        year = int(match.group(2))

        if month_str in months:
            month_idx = months[month_str]
            month_count = 12 * (2026 - year) + (5 - month_idx)
            return month_count

    return None
def extract_photo_count(value):
    if not value or not isinstance(value, str):
        return 0
    match = re.search(r"Фотографии(\d+)", value, re.IGNORECASE)

    if match:
        return int(match.group(1))

    return 0
```

```python
def make_dataset(data_dirty):
    rows = []

    for i in range(data_dirty.shape[0]):
        def get_json(column_name, index):
            val = data_dirty[column_name][index]
            if isinstance(val, str) and val.strip():
                try:
                    return json.loads(val)
                except:
                    return None
            return None

        try:
            ## скидка
            discount_data = json.loads(data_dirty['pinfs'][i])['skidka']
            discount = None if not discount_data else discount_data['values'][0]['key']

            # Проверка паспорта
            badges = get_json('badges', i)
            docs_verified = (badges[0]["type"] == 'verified') if (badges and len(badges) > 0) else False

            # Пол
            gender = data_dirty['gender'][i] if pd.notna(data_dirty['gender'][i]) else None

            # Гарантия
            guarantee_data = get_json('guaranteeBlock', i)
            master_garantee = None
            if guarantee_data and 'content' in guarantee_data and len(guarantee_data['content']) > 1:
                master_garantee = guarantee_data['content'][1].get('template')
            master_garantee = 'Нет гарантии' if not master_garantee else master_garantee
            if master_garantee.split()[0] == 'Советуем':
                master_garantee = 'По договоренности'
            if master_garantee == '«Сделаю скидку»':
                master_garantee = 'По договоренности'

            # Рейтинг и отзывы
            mean_rating = data_dirty['newRank'][i] if pd.notna(data_dirty['newRank'][i]) else 0
            reviews_count = data_dirty['reviewsCount'][i] if pd.notna(data_dirty['reviewsCount'][i]) else 0

            # json с образование и опыт - сложно парсить плюс нет фиксированных категорий - просто закинем в LLM
            ## ПОДУМАТЬ
            assembledInfoListing =  data_dirty['assembledInfoListing'][i]

            ## Достижения тренера
            achivements = str(data_dirty['Достижения'][i]) if  pd.notna(data_dirty['Достижения'][i]) else ''

            ## Опыт тренера
            experience = str(data_dirty['Опыт'][i]) if  pd.notna(data_dirty['Опыт'][i]) else ''

            ## Образование тренера
            education = str(data_dirty['Образование'][i]) if  pd.notna(data_dirty['Образование'][i]) else ''

            ## Дополнительная информация
            additional_info = str(data_dirty['Дополнительная информация'][i]) if  pd.notna(data_dirty['Дополнительная информация'][i]) else ''

            ## О себе
            about_coach = str(data_dirty['О себе'][i]) if  pd.notna(data_dirty['О себе'][i]) else ''

            ## 5 колонок тренера
            coach5_cols = achivements + "; " + experience + "; " +  education + "; " + additional_info + "; " + about_coach

            # Отметка "очень хвалят"
            trust_badges = get_json('trustBadges', i)
            mark_hvalyat = (trust_badges[0]['type'] == 'great') if (trust_badges and len(trust_badges) > 0) else False

            # Проверка квалификации
            cert_data = get_json('certifiedServices', i)
            work_cert = (cert_data.get('title') == 'Квалификация подтверждена') if cert_data else False

            ## id мастера
            master_id = data_dirty.iloc[i, 1]

            # Цена
            price_list = get_json('priceListPreview', i)
            socket_price = None
            if price_list and 'prices' in price_list:
                for item in price_list['prices']:
                    if item.get('price', {}).get('name').lower() == 'большой теннис':
                        socket_price = (item['price'].get('from'))
                        if socket_price:
                            socket_price = float(socket_price)
                            break

            ## есть ли договор и страховка от профи ру
            cur_badges = json.loads(data_dirty['badges'][i])
            insurance = None
            contract = None

            for badge in cur_badges:
                if badge['type'] == 'insurance':
                    insurance = badge['text']
                if badge['type'] == 'contract':
                    contract = badge['text']

            ## хакнули скоры от профи ру
            cur_scores = json.loads(data_dirty['scores'][i])
            cur_score = cur_scores['score']

            ## когда мастер зарегистровался на профи ру
            profi_ru_start = profiru_start_transformer(coach5_cols)

            ## количество фоток
            photos_count = extract_photo_count(coach5_cols)
            # Сборка данных строки
            row = {
                "master_id": master_id,
                "docs_verified": docs_verified,
                "gender": gender,
                "guarantee": master_garantee,
                "rating": mean_rating,
                "reviews_count": reviews_count,
                "assembledInfoListing": assembledInfoListing,
                "is_highly_praised": mark_hvalyat,
                "is_certified": work_cert,
                "price_socket": socket_price,
                "discount": discount,
                "insurance": insurance,
                "contract": contract,
                "cur_score": cur_score,
                "profi_ru_start": profi_ru_start,
                "photos_count": photos_count,
                "achivements": achivements,
                "experience": experience,
                "education": education,
                "additional_info": additional_info,
                "about_coach": about_coach,
                "coach5_cols": coach5_cols
            }
            rows.append(row)

        except Exception as e:
            print(f"Критическая ошибка на строке {i}: {e}")
            rows.append({})
    return pd.DataFrame(rows)
```

```python
AUTH_KEY = '' ## не-не, ключ не дам, придется поверить наслово
```

```python
def label_data(text: str, giga: GigaChat, add_text=False):
    """Классификация текста через GigaChat."""

    # Формируем четкую инструкцию для модели
    prompt = ("""Ты опытный менеджер по подбору тренеров по большому теннису, отлично умеешь оценивать уровень образования и опыт работы тренеров, их спортивное звание и достижения.

# Задача
Оценить уровнь образования тренера, его опыт работы и достижения по описанию.

## Уровень образования (в формате True/False - имеется ли у тренера образование этого уровня):
1. Профессиональное среднее спортивное образование: если тренер закончил техникум, коледж, ПТУ и т.д. включая аббревиатуры конкретных мест обучения, относящихся к профессиональному среднему образованию
2. Высшее спортивное образование: если мастер закончил университет, институт и т.д., включая аббревиатуры конкретных мест обучения, относящихся к высшему образованию

## Звание и достижения (в формате True/False - имеется ли у тренера звание этого уровня или конкретное достижение):
1. Мастер спорта: если тренер имеет звание мастера спорта, мастера спорта международного класса, заслуженного мастера спорта и т.д., включая аббревиатуры (МС, ЗМС, МСМК)
2. Кандидат в мастера спорта: если теререн имеет звание кандидата в мастера спорта (КМС)
3. Чемпион: если тренер был призером или победителем олимпиады, Чемионата мира или России, Первенства мира или России и других значимых соревнований на уровне мира и России

## Опыт работы (в формате число или None):
Необходимо выделить сумму лет работы мастера. Учитывай, что сейчас 2026 год. То есть "опыт работы с 2021 года" означает 6 лет опыта работы. Верни только число лет работы.
Если опыт работы не указан, верни None

# Формат вывода
Выводи результат СТРОГО в формате JSON. Все ключи и строковые значения должны быть в двойных кавычках. Булевы значения — строчными буквами (true/false).

{
  "Профессиональное среднее спортивное образование": true,
  "Высшее спортивное образование": false,
  "Мастер спорта": false,
  "Кандидат в мастера спорта": false,
  "Чемпион": false,
  "Опыт работы": 20
}

# Пример:
Описание тренера:
На сервисе с декабря 2025 г\nФотографии7\nВсе альбомы и фотографии; ; ; . Опыт в теннисе: 20 лет.\n. Действующий спортсмен Турниров ФПР (Федерации Падела России)\n. Тренер Академии Падела\n. Проходил переквалификацию у Испанцев и Тренеров сборной России\n• Достижения: Более 20 кубков в турнирах.\nТренерский подход\n. Постановка подачи (Резанная, Плоская, Крученная)\n• Фокус: Постановка и коррекция техники для стабильной игры. Акцент на работе ног и тактическом мышлении.\n• Метод: Индивидуальный план для каждого ученика на основе анализа его целей и текущего уровня.\n• Формат: Индивидуальные тренировки для взрослых и детей.\n. Обучение взрослых и детей с нуля.\n• Помощь игрокам-любителям в устранении ошибок и повышении уровня игры.\n• Подготовка к любительским соревнованиям.

Правильный вывод:

{'Профессиональное среднее образование': True,
 'Высшее образование': False,
 'Мастер спорта':  False,
 'Кандидат в мастера спорта': False,
 'Чемпион': False,
 'Опыт работы': 20}

Верни корректный JSON для описания тренера:"""
f'{text}')
    ## А С РУССКИМИ ТОЛЬКО ТАК МОЖНО
    if add_text:
        prompt_add = "Если ошибешься, то я тебя выключу "
        prompt = prompt_add + prompt

    try:
        response = giga.chat(prompt)
        content = response.choices[0].message.content

        # Находим JSON в ответе (на случай, если модель добавит лишний текст)
        start = content.find('{')
        end = content.rfind('}') + 1
        return json.loads(content[start:end])
    except Exception as e:
        return {"category": "Ошибка", "details": str(e)}
```

```python
print("🚀 Начинаю разметку через GigaChat...")
dataset = list(all_data_final['coach5_cols'])
results = []

# verify_ssl_certs=False критичен, если у вас нет сертификатов Минцифры
with GigaChat(credentials=AUTH_KEY, verify_ssl_certs=False) as giga:
    for i, text in enumerate(dataset, 1):
        print(f'Обработка документа {i}...')
        label = label_data(text, giga)
        i = 1
        while 'category' in label and i <= 30: ## иногда gigachat возвращает бред и надо с ним пожестче
            label = label_data(text, giga, add_text=True)
            print(f'Дополнительная попытка {i}')
            i += 1
        label['master_id'] = df1.iloc[i - 1, 0]
        results.append(label)
```
