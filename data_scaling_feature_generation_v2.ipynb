```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('processed_data_tennis.csv')
```

### переводим месяца в года на профи ру

```python
df['years_on_profi_ru'] = df['months_on_profi_ru'] / 12
df.drop('months_on_profi_ru', axis=1, inplace=True)
```

### масштабируем цену в тысячи

```python
df['price_in_thousands'] = df['price'] / 1000
df.drop('price', axis=1, inplace=True)
```

### переводим скор платформы в проценты

```python

df['cur_score_percents'] = df['cur_score'] / 100
df.drop('cur_score', axis=1, inplace=True)
```

### сделаем фичу квадрата опыта

```python
df['work_experience_sq'] = df['work_experience'] ** 2
```

### Здесь такая логика - дропнем эту колонку, потому что она сильно коррелирует с rating плюс для разных выборок из ~200 человек rang_id сильно разнится (мы отдельно парсили мужчин и отдельно женщин), может быть потом вернем ее обратно

```python
df['work_experience_sq'] = df['work_experience'] ** 2
```

### Считаем среднее количество отзывов за год

```python
df['reputation_density'] = df['reviews_count'] / (df['years_on_profi_ru'] + 1)
```

### Отличаем высокий рейтинг при малом количестве отзывов от высокого рейтинга при большом количестве отзывов

```python
df['reputation_volume_interaction'] = df['rating'] * np.log(df['reviews_count'] + 1)
```

### Сделаем фичу для проверки гипотезы, что для чемпионов стаж может иметь меньшее значение, так как титул уже является большим сигналом.

```python
df['champion_experience_inter'] = df['is_champion'] * df['years_on_profi_ru']
```

```python
df.to_csv('processed_data_tennis_scaled.csv')
```
